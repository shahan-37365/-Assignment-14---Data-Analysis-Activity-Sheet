```python
# Dataset 1: Pakistani E-Commerce Data
# Setup & Loading the Data
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Load the dataset
df_ecom = pd.read_csv(r'D:\ICA assignments\pakistani_ecommerce_dataset.csv')
# Quick look at the data
print(df_ecom.head())
print(df_ecom.info())
```

```python
# Q1: Most Popular Product Category
print(df_ecom['product_category'].value_counts())
```

```python
# Q2: Average Order Value
avg_order_value = df_ecom['order_value'].mean()
print(f'Average Order Value: {avg_order_value:.2f}')
print(df_ecom['order_value'].describe())
```

```python
# Q3: City with the Highest Number of Orders
print(df_ecom['city'].value_counts())
```

```python
# Q4: Most Used Payment Method
print(df_ecom['payment_method'].value_counts())
```

```python
# Q5: Bar Chart Showing Average Order Value by City
city_avg = (
    df_ecom.groupby('city')['order_value'].mean().sort_values(ascending=False)
)
print(city_avg)

plt.figure(figsize=(8, 5))
sns.barplot(
    x=city_avg.index, y=city_avg.values, palette='Blues_d', hue=city_avg.index
)
plt.title('Average Order Value by City')
plt.xlabel('City')
plt.ylabel('Average Order Value')
plt.tight_layout()
plt.show()
```

```python
# Dataset 2: Student Lifestyle & Study Habits
# Setup & Loading the Data
df_student = pd.read_csv(r'D:\ICA assignments\student_lifestyle_dataset.csv')
print(df_student.head())
print(df_student.info())
```

```python
# Q1: Average Number of Study Hours
avg_study = df_student['study_hours'].mean()
print(f'Average Study Hours: {avg_study:.2f}')
print(df_student['study_hours'].describe())
```

```python
# Q2: Who Sleeps More on Average — Males or Females?
sleep_by_gender = df_student.groupby('gender')['sleep_hours'].agg(
    ['count', 'mean', 'median', 'min', 'max', 'std']
)
print(sleep_by_gender)
```

```python
# Q3: Relationship Between Screen Time and Sleep Hours
correlation = df_student['screen_time'].corr(df_student['sleep_hours'])
print(f'Correlation between screen time and sleep hours: {correlation:.4f}')
print(df_student[['screen_time', 'sleep_hours']].describe())
```

```python
# Q4: Histogram of Daily Transport Cost
print(df_student['daily_transport_cost'].describe())

plt.figure(figsize=(8, 5))
plt.hist(
    df_student['daily_transport_cost'],
    bins=10,
    color='teal',
    edgecolor='black',
)
plt.title('Histogram of Daily Transport Cost')
plt.xlabel('Daily Transport Cost')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# Q5: Compare Average Internet Usage by Gender
internet_by_gender = df_student.groupby('gender')['internet_usage'].agg(
    ['count', 'mean', 'median', 'min', 'max', 'std']
)
print(internet_by_gender)
```
